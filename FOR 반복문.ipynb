```python
# for 와 range 사용하기
# for 변수 in range(횟수):
#     반복할 코드

for i in range(100):
    print("Hello World?")
```

```python
# 반복문에서 변수의 변화 알아보기

for i in range(10):
    print("hello, world", i)
```

```python
# 시작하는 숫자와 끝나는 숫자 지정하기

for a in range(5,12):
    print("Hello World!", a)
```

```python
for b in range(0, 10, 2):
    print("Hello, World!", b)
    
```

```python
for i in range(10, 0, -1):
    print("Hello, World!",i)
```
