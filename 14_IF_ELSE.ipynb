```python
# # else 사용하기
# # if 조건식:
#     수행할코드1
# else:
#     수행할코드2

x = 5

if x == 10:
    print("10입니다.")
else:
    print("10이 아닙니다.")
```

```python
# else와 들여쓰기

x = 5

if x == 10:
    print("10입니다.")
else:
    print("X에 들어있는 숫자는")
    print("10이 아닙니다.")
```

```python
# if 조건문의 동작방식 알아보기

if True:
    print("참")
else:
    print("거짓")

if False:
    print("참")
else:
    print("거짓")

if None:
    print("참")
else:
    print("거짓")
```

```python
# 조건식을 여러 개 지정하기

x = 10
y = 20

if x == 10 and y == 20:
    print("참")
else:
    print("거짓")
```
