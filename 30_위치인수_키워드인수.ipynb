```python
# 위치인수
def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)
print_numbers(10, 20, 30)
```

```python
# 언패킹 사용하기

x = [10, 20, 30]
print_numbers(*x)
```

```python
# # 가변 인수 함수 만들기
# def 함수이름(*매개변수):
#     코드

def print_numbers(*args):
    for arg in args:
        print(arg)
```

```python
print_numbers(10)
```

```python
print_numbers(10, 20, 30, 40)
```

```python
x = [10]
print_numbers(*x)
```

```python
y = [10, 20, 30, 40]
print_numbers(*y)
```

```python
# 키워드 인수 사용
def personal_info(name, age, address):
    print('이름: ', name)
    print('나이: ', age)
    print('주소: ', address)
personal_info(name='홍길동', age=30, address='서울시 용산구 이촌동')
```

```python
print(10, 20, 30, sep=':', end='')
```

```python
x = {'name': '홍길동', 'age': 30, 'address': '서울시 용산구 이촌동'}
personal_info(**x)
```

```python
# 가변 키워드 인수 함수 만들기

def personal_info(**kwargs):
    for kw, arg in kwargs.items():
        print(kw, arg, sep=" ")
personal_info(name = "홍길동")
```

```python
personal_info(name="홍길동", age= 30)
```

```python
# 매개변수에 초깃값 지정하기
def personal_info(name, age, address="비공개"):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
personal_info("홓길동", 30, "서울시 용산구")
```

```python
def personal_info(name, address="비공개", age):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```
