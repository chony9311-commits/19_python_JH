```python
#불과 비교 연산자 사용하기
print(True)
print(type(True))
print(False)
print(type(False))
```

```python
# 비교연산의 판단결과
print(3 > 1)
print(1 > 3)
print(3 < 1)
print(1 < 3)
print(3 == 3)
print(3 != 3)
```

```python
# 숫자가 같은지 다른지 비교
print(10 == 10)
print(10 != 10)
```

```python
print(10 < 20)
print(10 > 20)
print(10 <= 20)
print(10 >= 20)
print(10 <= 10)
print(10 >= 10)
```

```python
# 논리 연산자 사용하기 
True and True
print(True and True)
print(True and False)
print(False and False)
```

```python
True or True
print(True or True)
print(True or False)
print(False or False)

not True
print(not True)
print(not False)
```

```python
not True and False or not False
not True and False or not False
False and False or True
False or True
```

```python
# 논리 연상자와 비교 연산자를 함께 사용하기

print(10 == 10 and 10 != 5)
print(10 > 5 or 10 < 3)
print(not 10 > 5)
print(not 1 is 1.0)
```
