```python
import string

text = input()
words = text.split()

count = 0

for word in words:
    word = word.strip(string.punctuation)
    if word == 'the':
        count += 1

print(count)
```

```python

prices = list(map(int, input().split(';')))

prices.sort(reverse=True)

for price in prices:
    print(f'{price:>9,}')
```
