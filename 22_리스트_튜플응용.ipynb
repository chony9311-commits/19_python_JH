```python
# while 반복문으로 요소 출력하기
a = [38, 21, 53, 62, 19]

i = 0
while i < len(a):
    print(a[i])
    i += 1
```

```python
# 가장 작은 수와 가장 큰 수 구하기 
a = [38, 21, 53, 62, 19]
smallest = a[0]
for i in a:
    if i < smallest:
        smallest = i
print(smallest)
```

```python
a = [38, 21, 53, 62, 19]
largest = a[0]
for i in a:
    if i > largest:
        largest = i
print(largest)
```

```python
# 22.9 연습: 리스트에서 특정요소만 뽑아내기

a = ['alpha', 'bravo', 'charlie', 'delta', 'echo', 'foxtrot', 'golf', 'hotel', 'india']
b = [i for i in a if len(i) == 5]
print(b)
```

```python
# 22.10 심사: 2의 거듭제곱 리스트 생성하기
a, b = map(int, input().split())

x = [2 ** i for i in range(a,b + 1)]
del x[1]
del x[-2]
print(x)
```
