```python
# 변수 만들기

x = 10

# 영문 문자와 숫자를 사용할 수 있음
# 대소문자를 구분합니다.
# 문자부터 시작해야 하며 숫자로 시작할 수 없습니다.
# _(언더스코어)로 시작할 수 있습니다.
# 특수문자는 사용할 수 없습니다. (예: !, @, #, $, %, ^, &, *, (, ), -, +, =, {, }, [, ], |, \, :, ;, ", ', <, >, ,, ., ?, /)
# 파이썬의 키워드(예: if, else, for, while, def, class 등)는 변수 이름으로 사용할 수 없습니다.
```

```python
# 변수 여러개를 한꺼0번에 만들기

x, y, z = 10, 20, 30

print(x)
print(y)
print(z)
```

```python
a = input("첫번째 숫자를 입력하세요:")
b = input("두번째 숫자를 입력하세요:")
a = int(a)
b = int(b)
print(a+b)
```

```python
# 입력 값을 변수 두 개에 저장하기

a, b = input("문자열 두 개를 입력하세요:").split()
print(a)
print(b)
```

```python
# map을 사용하여 정수로 변환하기

a, b = map(int, input("숫자 두 개를 입력하세요:").split())
print(a+b)
```
