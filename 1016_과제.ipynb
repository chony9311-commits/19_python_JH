```python

students = [
    {"이름": "김민수", "국어": 85, "영어": 90, "수학": 78},
    {"이름": "이영희", "국어": 92, "영어": 88, "수학": 95},
    {"이름": "박철수", "국어": 78, "영어": 85, "수학": 82}
]

print("=== 학생 성적 관리 ===")

for student in students:
    total = student["국어"] + student["영어"] + student["수학"]
    average = total / 3

    student["총점"] = total
    student["평균"] = average

    print("이름:", student["이름"])
    print(f'국어: {student["국어"]}점, 영어: {student["영어"]}점, 수학: {student["수학"]}점')
    print(f'총점: {total}점, 평균: {average:.2f}점')
    print("---")
```

```python

text = "python is great python is fun python is powerful"

words = text.split()

count = {}

for word in words:
    if word in count:
        count[word] += 1
    else:
        count[word] = 1

for word, number in count.items():
    print(f"{word}: {number}번")
```

```python

school = {
    '1반': {
        '김철수': {'국어': 85, '영어': 90, '수학': 78},
        '이영희': {'국어': 92, '영어': 88, '수학': 95}
    },
    '2반': {
        '박민수': {'국어': 88, '영어': 85, '수학': 90},
        '최지영': {'국어': 95, '영어': 92, '수학': 88}
    }
}

print(school['1반']['김철수']['수학'])

scores = school['2반']['박민수']
average = sum(scores.values()) / len(scores)

print(f"{average:.2f}")
```
