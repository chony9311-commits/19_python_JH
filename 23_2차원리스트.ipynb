```python
# 반복문으로 2차원 리스트의 요소를 모두 출력하기

a = [[10, 20], [30, 40], [50, 60]]
for x, y in a:
    print(x, y)
```

```python
# for 반복문을 두 번 사용하기

a = [[10, 20], [30, 40], [50, 60]]
 
for i in a:        # a에서 안쪽 리스트를 꺼냄
    for j in i:    # 안쪽 리스트에서 요소를 하나씩 꺼냄
        print(j, end=' ')
    print()
```

```python
a = [[10, 20], [30, 40], [50, 60]]

for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j], end=" ")
    print()
```

```python
# whil 반복문을 한 번 사용하기

a = [[10, 20], [30, 40], [50, 60]]
i = 0 
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```
