```python
# 중첩 루프 사용하기
for i in range(5):
    for j in range(5):
        print("j:", j, sep="", end=" ")
        print('i:', i, '\\n', sep='')
```

```python
# 사각형으로 별 출력하기

for i in range(5):
    for j in range(5):
        print("*", end="")
    print()
```

```python
# 시각형 모양 바꾸기
for i in range(3):
    for j in range(7):
        print("*", end="")
    print()
```

```python
# 계단식으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j <= i:
            print("*", end="")
    print()
```

```python
# 대각선으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end="")
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
#19.5 연습: 역삼각형, 모양으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j < i:
            print(' ', end="")
        else:
            print("*", end="")
    print()
```

```python
#19.6 심사: 산 모양으로 별 출력하기

n = int(input())
for i in range(n):
    print(" " * (n-i-1) + "*" * (2 * i + 1))
```
