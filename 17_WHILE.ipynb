```python
# while 반복문 사용하기
# 초기식
# while 조건식:
#   반복할코드
#   변화식

i = 0
while i < 100:
    print("hello world")
    i += 1
```

```python
# 초기값을 감소시키기

i = 10
while i > 0:
    print("hello", i)
    i -= 1
```

```python
# 입력한 횟수대로 반복하기

count = int(input("반복할 횟수를 입력하세요: "))
i = 0
while i < count:
    print("Hello", i)
    i += 1
```

```python
# 반복횟수가 정해지지 않는 경우

count = int(input("반복할 횟수를 입력하세요: "))
while count > 0:
    print("Hello, world!", count)
    count -= 1
```
